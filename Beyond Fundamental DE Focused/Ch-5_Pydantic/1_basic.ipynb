```python
from pydantic import BaseModel
```

```python
class Employee_Schema(BaseModel): ###BaseModel have converted this class into a Pydantic Object
    name : str
    id :int
    department : str
    
    
```

```python
obj = Employee_Schema(name="Anish Shakya",id=10010,department="Data Analytics")
print(obj)
```

```python
obj2= Employee_Schema(**{"name":"Anish Shakya","id":100201,"department":"Data Engineering"})
obj2
```

```python
obj3= Employee_Schema(**{"name":"Anish Shakya","id":100201,"department":1201})
obj3
```

## **Coercion VS Strict Type**

### COERCION

```python
obj4= Employee_Schema(**{"name":"Anish Shakya","id":"100201","department":"Data Engineering"})
obj4

### here pydantic tried to convert the string value passed in id to integer but there is a limit
```

### StrictTypes

```python
from pydantic import StrictInt
```

```python
class employee_class(BaseModel):
    name:str
    id: StrictInt
    department : str
```

```python
obj = employee_class(name="Anish Shakya",id=123,department="Data")
obj
```
