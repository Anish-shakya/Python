```python
from pydantic import BaseModel,Field, EmailStr
from typing import Optional, Literal
import email_validator
```

```python
class Employee_Schema(BaseModel):
    name: str = Field(...,min_length=3,max_lenght=50)
    age: int = Field(...,gt=19,lt=65)
    email: EmailStr= Field(...,description="Employee's email address")
    department:Optional[str]=None
    gender: Literal['Male','Female','Other']=Field(...,description="Employee's Gender")
```

```python
obj = Employee_Schema(name="Anish Shakya",age='25',email='anish.shakya@omc.com',department='Data',gender='Male')
obj
```
