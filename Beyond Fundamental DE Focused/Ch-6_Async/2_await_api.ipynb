# **Syncronous API CALL**


```python
import time
```

```python

def Api_Call():
    print("Getting Data ....")
    time.sleep(3)
    return "Api Data"

def Execute():
    print("Executing API Call.......")
    response = Api_Call()
    print("Data Fetched:",response)
    
Execute()
    
```

# **Asyncronous API Call with coroutines**


```python
import asyncio

async def api_call(source):
    print(f"Getting Data from {source} API")
    await asyncio.sleep(5)
    return f"From {source}"

async def execute():
    print("Executing API Call.....")
    response = await api_call("Google Ads")
    print(f"Data Fectched Succesfully {response}")
    
    
asyncio.run(execute())



    
```

# **Asyncronous API Call With Multiple Coroutines**

```python
import asyncio

async def api_call(source):
    print(f"Getting Data from {source} API")
    await asyncio.sleep(5)
    return source

async def execute():
    print("Executing API Call.....")
    response = await asyncio.gather(api_call("Google Ads"),
                                    api_call("Meta Ads"),
                                    api_call("Tiktok Ads"),
                                    api_call("Snapchat Ads"))
    print(f"Data Fectched Succesfully From {response}")
    
asyncio.run(execute())
```
