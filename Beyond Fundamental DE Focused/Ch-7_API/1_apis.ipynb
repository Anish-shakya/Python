```python
import requests
import json
```

# **GET**


```python
url="https://jsonplaceholder.typicode.com/posts/1" 
response = requests.get(url) ## .get is the GET Method to request the data from the API

## Response Code is 200 means the status code. 200 means success the api call is perfect and response from API is received
print(response.status_code)

##URL
print(response.url)

## Headers
print(response.headers)
```

```python
# Response Body

print(response.text)
```

```python
# Response Body in json format and this is the recommended one

print(response.json())
```

```python
## parse json
data = json.dumps(response.json(),indent=2)
data
```

# **POST**


```python
url ="https://jsonplaceholder.typicode.com/posts"

data ={
    "title":"Learning API Post Method",
    "body":"Post Method is highly used in Data Engineering",
    "userId":1
}
```

```python
# Method 1: Using Json Parameter

response = requests.post(url,json=data) ## using the POST method to send the data to API , The json will automatiocally set the header for use
print(response.status_code)

print(json.dumps(response.json(),indent=2))

# Status Code 201 : 201 Means created or something is processed
```

```python
# Method 2: Using Form Data
form_data ={
    "title":"Learning API Post Method Using Form Data",
    "body":"Post Method is highly used in Data Engineering",
    "userId":2
}

response = requests.post(url,data=form_data)
print(response.status_code)

print(json.dumps(response.json(),indent=2))
```

# **Query Parameters**


```python
url ="https://jsonplaceholder.typicode.com/posts"

params={
    "userId":1,
    "id":1
}

response = requests.get(url,params=params) ### Its like a filter applied not a filter but lookalike
print(response.status_code)
print(json.dumps(response.json(),indent=2))

### After attaching the parameter the url looks like https://jsonplaceholder.typicode.com/posts?userId=1&id=1
```

```python
url ='https://jsonplaceholder.typicode.com/posts?userId=1&id=1'

response = requests.get(url)
print(json.dumps(response.json(),indent=2))
```

# **Headers and Authentication**

```python
url ="https://jsonplaceholder.typicode.com/posts"

headers = {
    "User-Agent":"python",
    "Accept":"application/json",
    "Custom-Header":"CustomValue"   
    
}

response =requests.get(url,headers=headers)
print(response.status_code)
print(response.headers)
##print(json.dumps(response.json(),indent=2))
```

```python
# Authentication JWT (Most recommeded for nowadays)

headers={
    "Authorization":f"Bearer {YOUR_ACCESS_TOKEN}",
    "Content-Type":"application/json"
}

# API Key Authentication

headers={
    "X-API-KEY":"YOUR_API_KEY"
}

# Basic Authentication (in postman as well we can use basic authentication)
from requests.auth import HTTPBasicAuth  

response = requests.get(url,auth=HTTPBasicAuth('username','password'))
```
