```python
def my_func(*x):
    return x

my_func(1,2,3,45)
```

# Lamda Functions



```python
def addition(x,y):
    return x+y

var = addition(10,20)
print(var)
```

```python
additions = lambda x,y: x+y
var2 = additions(20,10)
print(var2)
```

# Map, Filter and Reduce Function


```python
my_list=[1,2,3,4,5]

def square(x):
    ret_list =[i*i for i in x]
    print(ret_list)

square(my_list)    
```

```python
my_list=[1,2,3,4,5]

def square(p_x):
    return p_x*p_x

result = list(map(square,my_list))
result
```

```python
my_list=[1,2,3,4,5]

def square(p_x):
    if p_x%2==0:
        return p_x*p_x

result = list(filter(square,my_list))
result
```

```python
from functools import reduce

my_list=[1,2,3,4,5]

def square(p_x,p_y):
    return p_x+p_y

result = reduce(square,my_list)
result
```
