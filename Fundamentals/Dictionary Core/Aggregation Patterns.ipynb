#### Grouping with dictionary

```python
employees = [
    {"name": "Anish", "department": "Data"},
    {"name": "Rahul", "department": "Marketing"},
    {"name": "Sita", "department": "Data"},
    {"name": "Ram", "department": "Finance"},
    {"name": "Maya", "department": "Marketing"}
]

grouped_employees = {}

for employee in employees:
    department = employee['department']

    if department not in grouped_employees:
        grouped_employees[department]=[]   

    grouped_employees[department].append(employee['name'])
    
print(grouped_employees)


```

```python
sales = [
    {"product": "Laptop", "region": "East", "amount": 1000},
    {"product": "Monitor", "region": "West", "amount": 500},
    {"product": "Laptop", "region": "East", "amount": 1200},
    {"product": "Keyboard", "region": "West", "amount": 200}
]


## regional Sales
regional_sales ={}

for sale in sales:

    region = sale["region"]

    if region not in regional_sales:
        regional_sales[region]=[]

    regional_sales[region].append(sale)

regional_sales


##total sales by region
total_sales_region ={}
for sale in sales:
    region=sale["region"]

    if region not in total_sales_region:
        total_sales_region[region]=0
    
    total_sales_region[region]+=sale["amount"]

total_sales_region
```
