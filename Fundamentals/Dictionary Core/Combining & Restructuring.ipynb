### Dict as a Lookup Table (the "in-memory join") python

```python
customers = [
    {"id": 1, "name": "Alice"},
    {"id": 2, "name": "Bob"},
]

customer_by_id = {c['id']: c for c in customers}
print(customer_by_id[1])
```

```python
employee =[
    { 'emp_id' :'101', 'emp_name':'Anish Shakya','dept_id':'D01'},
    { 'emp_id' :'102', 'emp_name':'Sanish Shakya','dept_id':'D02'},
    { 'emp_id' :'103', 'emp_name':'Manish Shakya','dept_id':'D03'},
    { 'emp_id' :'104', 'emp_name':'Sanam Maharjan','dept_id':'D04'}]


dept_lookup ={
'D01':{'dept_name':'Data Engineering','location':'Kathmandu'},
'D02':{'dept_name':'Data Analytics','location':'Lalitpur'},
'D03':{'dept_name':'Marketing','location':'Bhaktapur'}
}
```

```python
result = []
for emp in employee:
    info = dept_lookup.get(emp['dept_id'],{'dept_name':'Unknown','location':'N/A'})
    combined = {**emp,**info}
    result.append(combined)

for i in result:
    print(i)
```

### Restructring and Renaming Keys

```python
customer = {
    "cust_id": 101,
    "cust_name": "Anish",
    "cust_email": "anish@gmail.com",
    "cust_country": "Nepal"
}

# But our data warehouse expects:
# customer_id
# customer_name
# email
# country

schema_mapping = {
    "cust_id":"customer_id",
    "cust_name":"customer_name",
    "cust_email":"email",
    "cust_country":"country"
}

new_customer ={}

for old_key,new_key in schema_mapping.items():
    new_customer[new_key]=customer[old_key]

print(new_customer)
```

```python
raw_data = {
    "customer_id": 101,
    "name": {
        "first": "Anish",
        "last": "Shakya"
    },
    "location": {
        "city": "Kathmandu",
        "country": "Nepal"
    }
}

restructure_schema = {
    "customer_id" :raw_data['customer_id'], 
    "customer_name":raw_data['name']['first']+" "+raw_data['name']['last'],
    "location":raw_data['location']['city'],
    "country":raw_data['location']['country']

}

print(restructure_schema)
```
