#### Dictionary to JSON



```python
import json

customer = {
    "customer_id": 101,
    "name": "Anish",
    "country": "Nepal"
}

json_data = json.dumps(customer)

print(json_data)
print(type(json_data))
```

#### Flattering Nested Dictionaries

```python
import json 

api_response = ''' { "id": 101, "name": "Anish Shakya", "username": "anish", "email": "anish@example.com", "address": { "street": "Main Road", "city": "Kathmandu", "zipcode": "44600", "geo": { "lat": "27.7172", "lng": "85.3240" } }, "company": { "name": "Example Analytics", "department": "Data Engineering" } } '''
```

```python
customer = json.loads(api_response)
```

```python
def flatten_dict(data, parent_key="", separator="_"):
    flattened = {}

    for key, value in data.items():

        new_key = f"{parent_key}{separator}{key}" if parent_key else key

        if isinstance(value, dict):
            flattened.update(
                flatten_dict(
                    value,
                    parent_key=new_key,
                    separator=separator
                )
            )
        else:
            flattened[new_key] = value

    return flattened
```

```python
flatten_dict(customer)
```

```python
import json 

api_response = ''' 
{
  "id": 101,
  "name": {
    "first_name": "Anish",
    "last_mame": "Shakya"
  },
  "username": "anish",
  "contact": {
    "email": "anish@example.com",
    "social_media": {
      "Meta": "facebook/anishshakya1",
      "Insta": {
        "id1": "instagram/anish1",
        "id2": "instagram/anishshaya2"
      }
    }
  },
  "address": {
    "Country": "Nepal",
    "Province": "Bagmati",
    "street": "Main Road",
    "city": "Kathmandu",
    "zipcode": "44600",
    "geo": {
      "lat": "27.7172",
      "lng": "85.3240"
    }
  },
  "company": {
    "name": "Example Analytics",
    "department": "Data Engineering"
  }
}

'''

customer = json.loads(api_response)

def flatten_dict(data, parent_key="", separator="_"):
    flattened = {}

    for key, value in data.items():

        new_key = f"{parent_key}{separator}{key}" if parent_key else key

        if isinstance(value, dict):
            flattened.update(
                flatten_dict(
                    value,
                    parent_key=new_key,
                    separator=separator
                )
            )
        else:
            flattened[new_key] = value

    return flattened

print(flatten_dict(customer))
```

```python
def flat_dictionary(raw_data,parent_key="",seperator="_"):
    
    flat_dict={}

    for key,value in raw_data.items():

        if parent_key:
            new_key = f'{parent_key}{seperator}{key}'
        else:
            new_key = key

        if isinstance(value,dict):
            flat_dict.update(
            flat_dictionary(value,parent_key=new_key,seperator=seperator))
        else:
            flat_dict[new_key]=value
    return flat_dict

flat_dictionary(customer)
```
