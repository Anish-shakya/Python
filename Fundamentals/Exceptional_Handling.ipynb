# Exception Handling


```python
x="10"

try:
    if(x>10):
        print("Greater than 10")
    else:
        print("Bla-Bla")
except Exception as e:
    print(f'hey, you got this error - {e}')
finally:
    print("I will always run")
print("Hello World!")
```

```python
def my_func(p_x):
    print("Hello Word")
    try:
        if(p_x%2==0):
            return 1
    except Exception as e:
        return e
    finally:  
        print("Hello Word")    

my_func(4)
```

# Global Variable In Function


```python
x=100 ##global variable

def my_func():
    x=5 ## local variable gets the priority
    print(x)

my_func()
```
