# List and List Comprehensions

```python
numbers = [4, 8, 15, 16, 23, 42]

even = [num for num in numbers if num%2==0]
print(even)
```

```python
words = ["data", "engineer", "sql", "python", "etl"]

lengths = [len(word) for word in words ]
print(lengths)
```

```python
temps_celsius = [0, 15, 30, 100, -10]

temps_fah = [i * 9/5 +32 for i in temps_celsius ]

print(temps_fah)
```

```python
records = [
    {"name": "Alice", "dept": "Engineering", "salary": 95000},
    {"name": "Bob", "dept": "Sales", "salary": 60000},
    {"name": "Carol", "dept": "Engineering", "salary": 110000},
    {"name": "Dave", "dept": "Marketing", "salary": 70000},
]


record = [ record['name'] for record in records if record['dept']=='Engineering' ]

print(record)
```

# Loops


```python
numbers = [12, 7, 19, 3, 25, 8]

large_number = numbers[0]

for num in numbers[1:]:
    if num > large_number:
        large_number=num

print(large_number)
```

```python
transactions = [100, -20, 50, -5, 300, -75]

total_credits =0
total_debits = 0

for amount in transactions:
    if amount<0 :
        total_debits-=amount
    else:
        total_credits+=amount

print(f'Total Debit {total_debits} And Total Credit {total_credits}')
```

```python
raw_records = ["10", "25", "abc", "40", "", "15"]
valid_total = 0

for record in raw_records:
    try:
        valid_total+=int(record)
    except ValueError:
        print(f"Skipping invalid record:{record}")
        continue
    
print(valid_total)
```

```python
fake_pages = {
    1: ["record_a", "record_b"],
    2: ["record_c"],
    3: [],  # empty page means no more data
}

page_number=1
has_more_data=True

while has_more_data:
    records=fake_pages[page_number]

    if not records:
        has_more_data =False
    else:
        print(f'Page {page_number}:{records}')
        page_number+=1
```

```python
product = {"name": "Laptop", "price": 1200, "stock": 5}

for key,value in product.items():
    print(f'{key}: {value}')
```

```python
inventory = {"apples": 50, "bananas": 0, "milk": 12, "eggs": 0}

zero_inventory ={ product:stock for product,stock in inventory.items() if stock>0}
print(zero_inventory)
```

```python
records = [
    {"name": "Alice", "age": 30},
    {"name": "Bob"},  # missing age — bad record
    {"name": "Carol", "age": 25},
]

for record in records:
    if 'age' in record:
        print(f"{record['name']} is {record['age']} years old.")
    else:
        print(f'Skipping {record['name']}: missing age')
```

```python
orders = [
    {
        "order_id": 101,
        "customer": {"name": "Alice", "country": "USA"},
        "items": [{"product": "Laptop", "price": 1200}, {"product": "Mouse", "price": 25}]
    },
    {
        "order_id": 102,
        "customer": {"name": "Bob", "country": "UK"},
        "items": [{"product": "Keyboard", "price": 45}]
    },
    {
        "order_id": 103,
        "customer": {"name": "Carol"},  # missing country — bad/incomplete record
        "items": []
    },
]
```

```python
for order in orders:
    name = order['customer']['name']
    country = order['customer'].get('country','Unknown')

    total_price = 0
    for items in order['items']:
        total_price+=items['price']
    
    print(f"Order {order['order_id']} - {name} ({country}) : ${total_price}")
```

```python
employees = [
    {
        "name": "Diana",
        "department": {"team": "Data Engineering", "manager": "Frank"},
        "projects": [
            {"project": "ETL Pipeline", "hours": 40},
            {"project": "Data Validation", "hours": 15},
        ]
    },
    {
        "name": "Ethan",
        "department": {"team": "Analytics"},  # missing "manager"
        "projects": [
            {"project": "Dashboard", "hours": 20},
        ]
    },
    {
        "name": "Fiona",
        "department": {"team": "Data Engineering", "manager": "Frank"},
        "projects": []  # no projects logged yet
    },
]
```

```python
for employee in employees:
    name = employee['name']
    department = employee['department']['team']
    manager =employee['department'].get('manager','Unassigned')

    total_hours =0

    for items in employee['projects']:
        total_hours+=items['hours']

    print(f"{name} ({department}, Manager: {manager}) - Total Hours: {total_hours}")

```
