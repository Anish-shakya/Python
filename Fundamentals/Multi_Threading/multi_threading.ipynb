```python
import time
import random
from concurrent.futures import ThreadPoolExecutor
```

```python
def my_func(p_x):

    wait = random.randint(1,10)
    time.sleep(wait)
    print(f'I am {p_x}. I took {wait} Seconds')

my_func(101)
```

```python
tables=['Orders','Products','Customers','Reviews','Cancelations']
```

```python
def my_processing(i):
    wait = 3
    time.sleep(wait)
    print(f'Processed {i}. It took {wait} Seconds')
```

```python
for i in tables:
    my_processing(i)
```

```python
#Using Multi Threading

with ThreadPoolExecutor(max_workers=len(tables)) as executor:

    futures = executor.map(my_processing,tables)
```

```python
with ThreadPoolExecutor(max_workers=len(tables)) as executor:
    for i in tables:
        future = executor.submit(my_processing,i)
```
