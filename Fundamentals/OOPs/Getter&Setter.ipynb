```python
class Employee():

    ## Default class attributed
    company_name ='XYZ'

    ##Dunder Method / Constructor 
    def __init__(self,emp_name,emp_dept):
    ## Instance attributes subject to the objects
        self.emp_name = emp_name
        self.emp_dept = emp_dept

    @property ## Getter Function
    def info(self):
        print(f"Employee {self.emp_name} works for {self.emp_dept} in {self.company_name}")    
    
    
    @info.setter ## Setter Function (This is the setter funtion for the getter info() )
    def info(self,new_empdetails):
        new_empname=new_empdetails[0]
        new_empdept=new_empdetails[1]
        
        self.emp_name=new_empname
        self.emp_dept=new_empdept
        

    #Methods / Alternative to class methods but not recommended
    def changes(self,new_company):
        Employee.company_name=new_company

    #Class Method
    @classmethod
    def changesInClass(cls,new_company):
        cls.company_name=new_company


    #Static method
    @staticmethod
    def addition(x,y):
        print(x+y)    




```

```python
emp1 = Employee("Anish","Data")
```

```python
##calling getter
print(emp1.info) ## need to call it like a atrribute
```

```python
emp1.info=['Anish Shakya','Data Analytics'] ## Calling Setter as an Attribute too
```
