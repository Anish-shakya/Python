# Single Level Inheritance

```python
class Company():

    def __init__(self,company_name,country):
        self.company_name =company_name
        self.country = country

    def company_info(self):
        print(f'Company Name is {self.company_name} in {self.country}')
    

class Employee(Company):

    def __init__(self,emp_name,company_name,country):
        self.emp_name=emp_name
        Company.__init__(self,company_name,country)
        ##self.company_name=company_name

    def emp_info(self):
        print(f"Employee Name is {self.emp_name}")   

    def company_info_child(self):
         print("This is Running From Employee Class")
         Company.company_info(self) ## works for all level

         ##Super().company_info() --> Accessing the the function from the parent class which is just one level above
         # this is only useful when working with single level inheritance. 
```

```python
emp1=Employee("Anish","XYZ","Nepal")
```

```python
emp1.emp_info()
emp1.company_info()
```

```python
emp1.company_info_child()
```

# Multiple Inheritance

```python
# Parent Class 1
class Company():

    def __init__(self,company_name):
        self.company_name =company_name

    def company_info(self):
        print(f'Company Name is {self.company_name}')

# Parent Class 2
class Country():

    def __init__(self,country_name,state_name):
        self.country_name=country_name
        self.state_name=state_name   
    
    def country_info(self):
        print(f"Country Name is {self.country_name} and located in {self.state_name} State")

# Child Class 
class Employee(Company,Country):

    def __init__(self,emp_name,emp_profession,company_name,country_name,state_name):
        self.emp_name=emp_name
        self.emp_profession=emp_profession
        Company.__init__(self,company_name)
        Country.__init__(self,country_name,state_name)
        
    def full_info(self):
        print(f"Employee {self.emp_name} Lives in {self.state_name} state of the {self.country_name}, and works for {self.company_name} as {self.emp_profession}")   

    def company_info_child(self):
         print("This is Running From Employee Class")
         Company.company_info(self)
         Country.country_info(self)

    
```

```python
emp1=Employee("Anish","Data Engineer","XYZ","Nepal","Bagmati")
```

```python
emp1.full_info()
```

# Multilevel Inheritance

```python
# Parent Class 1
class Company():

    def __init__(self,company_name):
        self.company_name =company_name

    def company_info(self):
        print(f'Company Name is {self.company_name}')

    def showme(self):
        print("I am Grandfather")

class Department(Company):
    
    def __init__(self,dept_name,company_name):
        self.dept_name=dept_name
        Company.__init__(self,company_name)

    def department_info(self):
        print(f"The department is {self.dept_name} of {self.company_name}")

    def showme(self):
        print("I am Father")
# Child Class 
class Employee(Department):

    def __init__(self,emp_name,emp_profession,company_name,dept_name):
        self.emp_name=emp_name
        self.emp_profession=emp_profession
        Company.__init__(self,company_name)
        Department.__init__(self,dept_name,company_name)
        
    def full_info(self):
        print(f"Employee {self.emp_name} works for {self.company_name} as {self.emp_profession} in {self.dept_name}")   

    def company_info_child(self):
         print("This is Running From Employee Class")
         Company.company_info(self)
         Department.department_info(self)

    def showme(self):
        print("I am a Pro")

    
```

```python
emp1=Employee("Anish","Data Engineer","XYZ","Data Analytics")
```

```python
emp1.full_info()
```

```python
emp1.company_info_child()
```

```python
emp1.company_info()
emp1.department_info()
```

```python
emp1.showme()
```
