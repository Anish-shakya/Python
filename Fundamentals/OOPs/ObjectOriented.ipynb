```python
class employee():

    ### Attributes
    emp_name="Anish"
    emp_dept="Pathways"

    ## Methods
    def info(self):
        print(f'Employee {self.emp_name} works for {self.emp_dept}')
    ## Method with parameter
    ## Here using a self means using a inner variable and using the the normal parameter means using external variable
    def info2(self,emp_name,emp_dept):
        print(f'Employee {self.emp_name} works for {emp_dept}')    
    ## Method with parameter
    def info3(self,emp_name,emp_dept):
        print(f'Employee {emp_name} works for {self.emp_dept}')    
    ## Method with parameter
    def info4(self,emp_name,emp_dept):
        print(f'Employee {emp_name} works for {emp_dept}')    

emp1 = employee()
print(emp1.emp_name)
print(emp1.emp_dept)
emp1.info()
emp1.info2("Anish Shakya","Data Engineer")
emp1.info3("Anish Shakya","Data Scientist")
emp1.info4("Anish Shakya","Data Analyst")

```

```python
class student:
    ## attributed
    def __init__(self,name,age,work):
        self.st_name = name
        self.st_age=age
        self.st_work=work

    ## method
    def st_info(self):
        print(f'Student {self.st_name} works in {self.st_work} at age of {self.st_age}')
```

```python
student1 = student("Anish",24,"Data Analyst")
student2 = student("Sanish",25,"Data Engineer")
student3 = student("Manish",26,"Data Scientist")

student1.st_info()
student2.st_info()
student3.st_info()
```

```python
class employee:
    ## Default class attributed
    emp_company ='XYZ'
   
    #Constructor
    def __init__(self,emp_name,emp_dept):
       ## Instance attributes subject to the objects
       self.emp_name=emp_name
       self.emp_dept=emp_dept
    ## Methods
    def info(self):
        print(f'Employee {self.emp_name} works for {self.emp_dept} at {self.emp_company}')

emp1 = employee("Anish Shakya","Data")
emp1.info()
```

```python
class bank():

    bank_name ="Sidddartha Bank"

    def __init__(self,account_holder,account_no,account_type,balance=0.0):
        self.account_holder=account_holder
        self.account_no = account_no
        self.account_type= account_type
        self.balance = balance
        self.account_statement = []

    def deposit(self,amount):
        try:
            if amount < 1:
                raise Exception("The deposit amount cannot be less that 1 or negative")     
            else:
                self.balance+=amount
                self.account_statement.append(f'Deposited Rs.{amount} by {self.account_holder}')
        except Exception as e:
            print(e)

    def withdrawal(self,amount):
        try:
            if amount < 0:
                raise Exception("The withdrawal amount should be greater than 0")
            elif amount <= self.balance:
                self.balance-=amount
                self.account_statement.append(f"Withdrawal Rs.{amount} by {self.account_holder}")
            else:
                raise Exception("Your Account Has Insufficient Balance")
        except Exception as e:
            print(f'{e}')   
               
    def transaction(self,benificiary,amount):
        try:
            if amount<=0:
                raise Exception('Transfer amount should be greater than 0')
            
            elif self.balance>=amount :
                self.withdrawal(amount)
                benificiary.deposit(amount)

                self.account_statement.pop()
                self.account_statement.append(f'Fund Transfer to {benificiary.account_holder} Rs.{amount}')

                benificiary.account_statement.pop()
                benificiary.account_statement.append(f'Fund Deposited by {self.account_holder} Rs.{amount}')
            else:
                raise Exception('Your account has insufficient balance')    

        except Exception as e:
            print(e)    

    def getBalance(self):
        return self.balance
    
    def info(self):
        print('----------------------------------')
        print(f'Bank Name : {self.bank_name}')
        print(f'Account Holder : {self.account_holder}')
        print(f'Account No : {self.account_no}')
        print(f'Account Account Type : {self.account_type}')
        print(f'Account Balance : {self.balance}')
        print('----------------------------------')

    def statement(self):
        print(f'------------{self.bank_name}--------------')
        print(f'Account Statement For : {self.account_holder}')
        for i,x in enumerate(self.account_statement,start=1):
            print('---------------------------------------')
            print(f'{i} : {x}')
           
account1 = bank('Anish Shakya','123456789','Saving Account',1000)
account2 = bank('Sanam Maharjan','987654321','Saving Account',5000)
```

```python
account1.deposit(1000)
account2.deposit(1000)
print(account1.balance)
print(account2.balance)
```

```python
account1.withdrawal(500)
account2.withdrawal(500)
print(account1.balance)
print(account2.balance)
```
