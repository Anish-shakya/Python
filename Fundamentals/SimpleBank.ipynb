```python
from datetime import datetime

# ============================================================ 
#  BANK ACCOUNT CLASS
#  ============================================================

class BankAccount():

    bank_name='Global IME Bank'

    def __init__(self,account_holder,account_no,account_type,balance=0.0):
        if balance < 0: 
            raise ValueError("Initial balance cannot be negative")
        self.account_holder = account_holder
        self.account_no = account_no
        self.account_type=account_type
        self.balance=balance
        self.transactions=[]

        self.transactions.append({
            "type":"Deposit",
            "amount":f"Rs.{balance}",
            "date":datetime.now(),
            "balance_after":balance,
            "description":f'Opening Balance'
        })
        
# ============================================================ 
#  DEPOSIT
#  ============================================================        

    def deposit(self,amount):
        ## validate deposit amount
        if amount<=0:
            raise ValueError("Deposit Amount Should Be Greater Than 0")
        
        self.balance+=amount

        ## Maintain Statement
        self.transactions.append({
            "type":"Deposit",
            "amount":f"Rs.{amount}",
            "date":datetime.now(),
            "balance_after":self.balance,
            "description":f'Deposited by {self.account_holder}'
        })
# ============================================================ 
#  WITHDRAWAL
#  ============================================================
    def withdrawal(self,amount):
        ## validate withdrawal
        if amount<=0:
            raise ValueError('Withdrawal Amount Cannot Be Less Than 0')    

        self.balance-=amount

        ## Maintain Statement
        self.transactions.append({
            "type":"Withdraw",
            "amount":f"Rs.{amount}",
            "date":datetime.now(),
            "balance_after":self.balance,
            "description":f'Withdrawal by {self.account_holder}'
        })
# ============================================================ 
#  FUND TRANSFERS
#  ============================================================
    def transfer(self,benificiary,amount):
        ## Validate amount

        if amount<=0:
            raise ValueError("Transfer Amount Must Be Greater Than 0")
        
        ## check for Sufficient Balance
        if amount>self.balance:
            raise ValueError("Your Account Has Insufficient Balance")
        
        ## Preventing transfer to own account
        if self.account_no == benificiary.account_no:
            raise ValueError("Cannot Transfer Fund To Same Account")
        
        #Deduct amount from sender account
        self.balance -= amount

        #Add amount to benificiary account
        benificiary.balance+=amount

        ##Maintain Sender Statement
        self.transactions.append({
            "type":"Fund Transfer",
            "amount":f"Rs.{amount}",
            "date":datetime.now(),
            "balance_after":self.balance,
            "description":f'Fund Transfer Rs.{amount} to {benificiary.account_holder}' 
        })

          ##Maintain Receiver Statement
        benificiary.transactions.append({
            "type":"Fund Received",
            "amount":f"Rs.{amount}",
            "date":datetime.now(),
            "balance_after":benificiary.balance,
            "description":f'Fund Received Rs.{amount} From {self.account_holder}' 
        })

# ============================================================ 
#  BALANCE CHECK
#  ============================================================
    def get_balance(self):
        return self.balance
# ============================================================ 
#  ACCOUNT INFO
#  ============================================================    
    def info(self):
        print() 
        print("------------------------------------------") 
        print(f"Bank Name : {self.bank_name}") 
        print(f"Account Holder : {self.account_holder}") 
        print(f"Account No : {self.account_no}") 
        print(f"Account Type : {self.account_type}") 
        print(f"Account Balance: Rs.{self.balance}") 
        print("------------------------------------------")
# ============================================================ 
#  BANK STATEMENT
#  ============================================================
    def statement(self):
        print()
        print(f"------------ {self.bank_name} ------------")
        print(f"Account Statement: {self.account_holder}")
        print()

        if not self.transactions:
            print("No Transactions Available.")
            return

        # --------------------------------------------------------
        # TABLE HEADER
        # --------------------------------------------------------

        print("-" * 150)

        print(
            f"{'Transaction No':<17}"
            f"{'Date':<22}"
            f"{'Transaction Type':<20}"
            f"{'Transaction Amount':<20}"
            f"{'Description':<45}"
            f"{'Account Balance':<25}"
        )

        print("-" * 150)

        # --------------------------------------------------------
        # TABLE DATA
        # --------------------------------------------------------

        for index, transaction in enumerate(self.transactions,start=1):
            print(
                f"{index:<17}"
                f"{transaction['date'].strftime('%Y-%m-%d %H:%M:%S'):<22}"
                f"{transaction['type']:<20}"
                f"{transaction['amount']:<20}"
                f"{transaction['description']:<45}"
                f"Rs.{transaction['balance_after']:<25}"
            )
        print("-" * 150)
            
            
```

```python
Anish=BankAccount('Anish Shakya','123456789','Saving Account',1500)
Sanam=BankAccount ('Sanam Maharjan','987654321','Saving Account',2000)   
Suman=BankAccount ('Suman Shrestha','741852963','Saving Account',3000)   
Julia=BankAccount ('Julia Dhakal','147258369','Saving Account',5000)   
```

```python
print(Anish.get_balance())
print(Sanam.get_balance())
print(Suman.get_balance())
print(Julia.get_balance())
```
