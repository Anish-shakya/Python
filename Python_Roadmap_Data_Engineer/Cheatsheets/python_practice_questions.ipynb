# List Comprehension Practice Questions

1. You're analyzing weather data. Given a list of temperatures, filter out all temperatures below 15°C and create a new list using list comprehension. Input: [5, 20, 12, 25, 8, 30, 3, 18]

```python
temp_list = [5, 20, 12, 25, 8, 30, 3, 18]

temp_above_15 =[i for i in temp_list if i>15]
print(temp_above_15)
```

2. An e-commerce platform filters product names by length. Given a list of product names, create a list containing only names with more than 5 characters using list comprehension. Input: ['phone', 'keyboard', 'mouse', 'monitor', 'usb']

```python
item_list = ['phone', 'keyboard', 'mouse', 'monitor', 'usb']
item_list_5ch = [item for item in item_list if len(item)>5]

print(item_list_5ch)
```

3. A restaurant wants to calculate ingredient costs. Given quantities (in kg), calculate total cost where each kg costs as its square value. Use list comprehension. Input: [2, 3, 5, 1, 4]

```python
my_list = [2, 3, 5, 1, 4]
cost = [i*i for i in my_list]

print(cost)
```

4. A password validator extracts uppercase letters to check password strength. Given a string, extract all uppercase letters using list comprehension. Input: 'HeLLo_World_123'

```python
password = 'HeLLo_World_123'
my_list = list(password)

upper_list = [letter for letter in my_list if letter.isupper()]
print(upper_list)
```

5. HR system filters eligible candidates for senior roles (age 30+). Given (name, age) tuples, extract names of eligible people using list comprehension. Input: [('Alice', 28), ('Bob', 32), ('Charlie', 25), ('Diana', 35)]

```python
my_list = [('Alice', 28), ('Bob', 32), ('Charlie', 25), ('Diana', 35)]

eligible = [name for name,age in my_list if age>=30]

print(eligible)
```

7. Calculate monthly revenue. Given a list of daily sales, calculate total revenue for the month using a loop. Input: [1200, 1500, 980, 2100, 1300, 1800, 900]

```python
daily_sales = [1200, 1500, 980, 2100, 1300, 1800, 900]
total_sales = 0
for sales in daily_sales:
    total_sales+=sales

print(f'Total Sales of the month {total_sales}') 
```

8. Log analyzer reads system errors in reverse chronological order. Given a list of log messages, print each from last to first. Input: ['Login error', 'DB timeout', 'File not found', 'Connection refused']

```python
log_list = ['Login error', 'DB timeout', 'File not found', 'Connection refused']

for logs in log_list[::-1]:
    print(logs)
    #########################
    
for logs in reversed(log_list):
    print(logs)    
```

9. Banking app filters successful transactions. Given transaction amounts (positive=credit, negative=debit), print only successful deposits using a loop. Input: [500, -200, 1500, -50, 2000, -100, 300]

```python
transactions = [500, -200, 1500, -50, 2000, -100, 300]

for transaction in transactions:
    if transaction > 0:
        print(transaction)
```

10. Website analytics finds peak traffic hour. Given hourly traffic numbers, find the hour with max traffic using a loop (no max() function). Input: [150, 320, 280, 450, 210, 380, 290, 500, 180]

```python
traffic = [150, 320, 280, 450, 210, 380, 290, 500, 180]

highest_traffic = traffic[0]
hour = 0

for i in range(1,len(traffic)):
    if traffic[i]>highest_traffic:
        highest_traffic = traffic[i]
        hour = i


print(f"Hour {hour} and traffic {highest_traffic}")


```

16. University processes exam results. Given student scores dict, print names scoring 85+. Input: {'Alice': 92, 'Bob': 78, 'Charlie': 88, 'Diana': 81, 'Eve': 95}

```python
my_dict ={'Alice': 92, 'Bob': 78, 'Charlie': 88, 'Diana': 81, 'Eve': 95}

for key,value in my_dict.items():
    if value > 85:
        print(key)
```

18. Grocery store generates stock report. Given item_name: quantity dict, print all stock quantities. Input: {'apples': 45, 'oranges': 30, 'bananas': 60, 'grapes': 25}

19. Electronics store finds premium item. Given product: price dict, find the most expensive item. Input: {'laptop': 1200, 'mouse': 45, 'keyboard': 120, 'monitor': 350, 'headphones': 95}

```python
products={'laptop': 1200, 'mouse': 45, 'keyboard': 120, 'monitor': 350, 'headphones': 95}

most_expensive = max(products,key=products.get)
print(most_expensive)
```
